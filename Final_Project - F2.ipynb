# 1. Project Overview

## 1.1 Overview of Banking in Portugal

Bank merupakan institusi keuangan yang menyediakan berbagai produk dan layanan untuk memenuhi kebutuhan finansial nasabah, baik untuk menyimpan dana, melakukan transaksi pembayaran, maupun memperoleh pembiayaan.

Portugal memiliki sektor perbankan yang menjadi bagian penting dari sistem keuangan nasional dan terintegrasi dengan sistem keuangan kawasan euro. Bank berperan sebagai lembaga intermediasi yang menghimpun dana dari masyarakat, terutama melalui deposito, kemudian menyalurkan dana tersebut melalui pemberian kredit dan investasi pada aset keuangan. Selain itu, bank juga menyediakan berbagai layanan pembayaran dan layanan keuangan lainnya.

Kondisi sektor perbankan Portugal juga menunjukkan perkembangan positif dalam beberapa aspek. Pada tahun 2024, profitabilitas sektor perbankan tetap tinggi, sementara kualitas kredit relatif stabil, dengan rasio non-performing loans (NPL) sekitar 2,4%. Kondisi tersebut menunjukkan bahwa aktivitas perbankan tidak hanya berkaitan dengan penyaluran kredit, tetapi juga mencakup pengelolaan dana masyarakat, pengelolaan risiko kredit, dan pemeliharaan stabilitas keuangan.
Referensi: Banco de Portugal – Banking Sector. (https://bpstat.bportugal.pt/api/media/files/scripts/numerosquecontam/2853/en/conteudo_setorbancario.html)

Dalam konteks tersebut, produk deposito menjadi salah satu aktivitas penting dalam perbankan, karena memungkinkan masyarakat menyimpan dana pada lembaga kredit berdasarkan ketentuan dan imbal hasil yang telah disepakati. Salah satu bentuk produk deposito adalah term deposit (deposito berjangka), yaitu deposito dengan jangka waktu tertentu yang menjadi fokus produk dalam proyek ini.

## 1.2 Produk Utama Bank

Secara umum, produk perbankan dapat dikelompokkan ke dalam beberapa kategori utama:

**1. Deposit Products**

Produk deposito digunakan oleh nasabah untuk menyimpan dana pada bank. Jenis deposito dapat dibedakan berdasarkan tingkat likuiditas, jangka waktu, dan ketentuan pencairannya. Banco de Portugal menjelaskan beberapa bentuk deposito, termasuk current accounts, deposits redeemable at notice, time deposits, dan special-regime deposits.

**2. Credit Products**

Bank juga menyediakan produk pembiayaan seperti kredit konsumen dan kredit perumahan. Produk ini memungkinkan nasabah memperoleh dana dari bank dengan kewajiban pembayaran kembali sesuai ketentuan kontrak.

**3. Payment and Transaction Services**

Bank menyediakan rekening pembayaran dan berbagai instrumen transaksi seperti kartu pembayaran, transfer, direct debit, dan layanan perbankan lainnya. Current account merupakan salah satu contoh rekening yang memungkinkan nasabah melakukan penyimpanan dan pergerakan dana untuk kebutuhan transaksi.

Dengan berbagai kategori tersebut, bank tidak hanya berfungsi sebagai tempat penyimpanan dana, tetapi juga menyediakan layanan transaksi, pembiayaan, dan berbagai produk keuangan lainnya

## 1.3 Term Deposit as the Focus Product

Salah satu produk deposito yang menjadi fokus dalam project ini adalah term deposit, atau deposito berjangka.

Term deposit merupakan produk simpanan di mana dana ditempatkan pada bank untuk jangka waktu tertentu sesuai dengan ketentuan yang disepakati. Berbeda dengan current account yang dapat digunakan untuk transaksi dan penarikan dana secara lebih fleksibel, dana pada time deposit pada umumnya ditempatkan hingga akhir periode yang telah ditentukan. ## 2.8 Confusion Matrix

Karena target pada dataset memiliki distribusi kelas yang tidak seimbang, penggunaan Accuracy saja belum cukup untuk menggambarkan kemampuan model dalam mengidentifikasi nasabah yang melakukan subscription. Oleh karena itu, digunakan Confusion Matrix untuk melihat hasil klasifikasi secara lebih rinci berdasarkan empat kategori, yaitu True Positive (TP), True Negative (TN), False Positive (FP), dan False Negative (FN).

<div style="border:2px solid black; padding:10px; width:50%; text-align:center;"> <table> <tr> <th></th><th>Predicted: No</th><th>Predicted: Yes</th> </tr> <tr> <th>Actual: No</th><td>True Negative (TN)</td><td>False Positive (FP)</td> </tr> <tr> <th>Actual: Yes</th><td>False Negative (FN)</td><td>True Positive (TP)</td> </tr> </table> </div> 
<br>

**True Negative (TN):**
Nasabah diprediksi tidak melakukan subscription dan aktualnya tidak melakukan subscription.

**False Positive (FP):** 
Nasabah diprediksi melakukan subscription tetapi aktualnya tidak melakukan subscription.

**False Negative (FN):**
Nasabah diprediksi tidak melakukan subscription tetapi aktualnya melakukan subscription.

**True Positive (TP):** 
Nasabah diprediksi melakukan subscription dan aktualnya melakukan subscription.


Dalam penelitian ini, F2-Score digunakan sebagai metric utama, sedangkan Recall dan Precision digunakan sebagai metric pendukung. Pemilihan F2-Score dilakukan karena tujuan model lebih menekankan kemampuan dalam mengidentifikasi nasabah yang benar-benar melakukan subscription. F2-Score memberikan bobot yang lebih besar terhadap Recall dibandingkan Precision, sehingga kesalahan berupa False Negative (FN) menjadi perhatian penting dalam evaluasi model.

## 2.9 Evaluation Metric

Dalam penelitian ini, evaluasi model dilakukan dengan menggunakan beberapa metrik, yaitu F2-Score, Recall, Precision, F1-Score, ROC-AUC, dan PR-AUC.

Pemilihan metrik disesuaikan dengan karakteristik dataset yang memiliki distribusi kelas tidak seimbang serta tujuan utama penelitian, yaitu mengidentifikasi nasabah yang berpotensi melakukan subscription.

**Primary
F2-Score**

F2-Score digunakan sebagai metric utama karena memberikan bobot lebih besar terhadap Recall dibandingkan Precision. Dengan demikian, evaluasi model lebih menekankan kemampuan model dalam mengidentifikasi nasabah yang benar-benar melakukan subscription, namun tetap mempertimbangkan ketepatan prediksi positif.

Formula F2-Score:

$$ F_2 = (1+2^2)\times \frac{Precision\times Recall} {(2^2\times Precision)+Recall} $$

atau:

$$ F_2 = 5\times \frac{Precision\times Recall} {4\times Precision+Recall} $$

F2-Score digunakan untuk memilih model dan threshold yang memberikan keseimbangan antara kemampuan menangkap subscriber dan ketepatan dalam menentukan target marketing, dengan penekanan yang lebih besar pada Recall.

**Secondary**

**Recall**

Recall digunakan sebagai metric pendukung untuk mengukur kemampuan model dalam mengidentifikasi nasabah yang benar-benar melakukan subscription.

$$ Recall = \frac{TP}{TP+FN} $$

Recall menjawab pertanyaan:

Dari seluruh customer yang sebenarnya melakukan subscription, berapa banyak yang berhasil diidentifikasi oleh model?

Recall menjadi metric yang penting karena False Negative (FN) menunjukkan nasabah yang sebenarnya melakukan subscription tetapi tidak berhasil diidentifikasi oleh model.

**Precision**

Precision digunakan sebagai metric pendukung untuk mengukur ketepatan model dalam menentukan nasabah yang diprediksi akan melakukan subscription.

$$ Precision = \frac{TP}{TP+FP} $$

Precision menjawab pertanyaan:

Dari customer yang diprediksi akan melakukan subscription, berapa banyak yang benar-benar melakukan subscription?

Precision digunakan untuk melihat potensi ketepatan targeting marketing dan mengontrol jumlah False Positive (FP), yaitu nasabah yang diprediksi akan melakukan subscription tetapi ternyata tidak melakukan subscription.

**Supporting**

**F1-Score** digunakan sebagai metric tambahan untuk mengukur keseimbangan antara Precision dan Recall dengan memberikan bobot yang sama terhadap keduanya.

$$ F_1 = 2\times \frac{Precision\times Recall} {Precision+Recall} $$

F1-Score digunakan sebagai pembanding terhadap F2-Score. Perbedaannya adalah F1 memberikan bobot yang sama terhadap Precision dan Recall, sedangkan F2 memberikan bobot yang lebih besar terhadap Recall.

**ROC-AUC** digunakan untuk mengukur kemampuan model dalam membedakan antara nasabah yang melakukan subscription dan nasabah yang tidak melakukan subscription pada berbagai nilai threshold.

ROC-AUC memberikan gambaran mengenai kemampuan diskriminasi model secara keseluruhan tanpa bergantung pada satu threshold tertentu.


**PR-AUC** digunakan untuk mengevaluasi hubungan antara Precision dan Recall pada berbagai threshold.

Metrik ini digunakan sebagai metrik tambahan karena dataset memiliki distribusi kelas yang tidak seimbang, sehingga evaluasi terhadap positive class, yaitu nasabah yang melakukan subscription, menjadi penting.

**Dengan pendekatan tersebut, model diarahkan untuk memaksimalkan kemampuan mengidentifikasi calon subscriber dengan tetap mempertimbangkan ketepatan targeting marketing.**

```python
# Outlier Summary
outlier_summary = []

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower,
        "Upper Bound": upper,
        "Outlier Count": outliers,
        "Outlier %": outliers / len(df) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df.sort_values(
    "Outlier %",
    ascending=False
)
```

## Step 14 - Customer Demographic Analysis vs Subscription

```python
# subscription rate per group.
def subscription_rate_by(df, column):

    result = (
        df.groupby(column)["y"]
        .agg(
            total_customers="count",
            subscribers=lambda x: (x == "yes").sum()
        )
    )

    result["subscription_rate"] = (
        result["subscribers"] /
        result["total_customers"] * 100
    )

    return result.sort_values(
        "subscription_rate",
        ascending=False
    )
```

## Step 15 - Age Group

```python
# ============================================================
# AGE GROUP
# ============================================================

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=[
        "<=25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        ">65"
    ]
)

# Subscription rate
age_subscription = subscription_rate_by(
    df,
    "age_group"
)

age_plot = age_subscription.reset_index()


# ============================================================
# Total Customer
# ============================================================

age_count = (
    df["age_group"]
    .value_counts()
    .sort_index()
    .reset_index()
)

age_count.columns = [
    "age_group",
    "total_customers"
]


# ============================================================
# 2 BAR CHART BERDAMPINGAN
# ============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 5)
)


# ------------------------------------------------------------
# Chart 1 — Subscription Rate
# ------------------------------------------------------------

ax1 = sns.barplot(
    data=age_plot,
    x="age_group",
    y="subscription_rate",
    ax=axes[0]
)

axes[0].set_title(
    "Subscription Rate by Age Group"
)
axes[0].set_xlabel("Age Group")
axes[0].set_ylabel("Subscription Rate (%)")

for p in ax1.patches:

    height = p.get_height()

    axes[0].annotate(
        f"{height:.2f}%",
        (
            p.get_x() + p.get_width() / 2,
            height
        ),
        ha="center",
        va="bottom",
        fontsize=10,
        xytext=(0, 5),
        textcoords="offset points"
    )


# ------------------------------------------------------------
# Chart 2 — Total Customer
# ------------------------------------------------------------

ax2 = sns.barplot(
    data=age_count,
    x="age_group",
    y="total_customers",
    ax=axes[1]
)

axes[1].set_title(
    "Total Customer by Age Group"
)
axes[1].set_xlabel("Age Group")
axes[1].set_ylabel("Total Customer")

for p in ax2.patches:

    height = p.get_height()

    axes[1].annotate(
        f"{int(height):,}",
        (
            p.get_x() + p.get_width() / 2,
            height
        ),
        ha="center",
        va="bottom",
        fontsize=10,
        xytext=(0, 5),
        textcoords="offset points"
    )


plt.tight_layout()
plt.show()
```

## Step 5 - Define Evaluation Metrics

```python
def evaluate_model(
    model_name,
    y_true,
    y_pred,
    y_prob
):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    # F2-Score
    f2 = fbeta_score(
        y_true,
        y_pred,
        beta=2,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_true,
        y_prob
    )

    pr_auc = average_precision_score(
        y_true,
        y_prob
    )

    print("\n" + "=" * 60)
    print(model_name)
    print("=" * 60)

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-Score : {f1:.4f}")
    print(f"F2-Score : {f2:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")
    print(f"PR-AUC   : {pr_auc:.4f}")

    print("\nConfusion Matrix:")
    print(
        confusion_matrix(
            y_true,
            y_pred
        )
    )

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "F2": f2,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc
    }
```

## Step 6 - Modelling

### Step 6A — Define Feature Groups

```python
target_encode_cols = [
    "job",
    "education_group",
    "marital",
    "month",
    "poutcome"
]

onehot_cols = [
    "default",
    "housing",
    "loan",
    "contact",
    "day_of_week",
    "age_group",
    "campaign_group"
]

# All remaining columns are numeric
numeric_cols = [
    col for col in X_train.columns
    if col not in target_encode_cols + onehot_cols
]

print("Target Encoding columns:")
print(target_encode_cols)

print("\nOne-Hot Encoding columns:")
print(onehot_cols)

print("\nNumeric columns:")
print(numeric_cols)
```

### Step 6B — Create Preprocessing Pipeline

```python
def create_preprocessor():

    preprocessor = ColumnTransformer(
        transformers=[

            (
                "target_encoding",
                TargetEncoder(
                    target_type="binary",
                    smooth="auto",
                    cv=5,
                    shuffle=True,
                    random_state=42
                ),
                target_encode_cols
            ),

            (
                "onehot",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                ),
                onehot_cols
            ),

            (
                "numeric",
                "passthrough",
                numeric_cols
            )
        ],

        remainder="drop"
    )

    return preprocessor
```

### Step 6C — Function untuk Membuat Model Pipeline

```python
def create_model_pipeline(
    model,
    use_smote=False,
    use_scaler=False
):

    steps = [
        (
            "preprocessor",
            create_preprocessor()
        )
    ]

    if use_scaler:

        steps.append(
            (
                "scaler",
                RobustScaler()
            )
        )

    if use_smote:

        steps.append(
            (
                "smote",
                SMOTE(random_state=42)
            )
        )

    steps.append(
        (
            "model",
            model
        )
    )

    return ImbPipeline(steps)
```

## Step 10 - Tabel Hasil Cross Validation

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    roc_auc_score,
    average_precision_score
)
from catboost import CatBoostClassifier
import pandas as pd


# ============================================================
# STRATIFIED 5-FOLD CV
# ============================================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# Menyimpan hasil setiap fold
catboost_cv_results = []


# ============================================================
# MANUAL CROSS-VALIDATION
# ============================================================

for fold, (train_idx, val_idx) in enumerate(
    cv.split(X_train, y_train),
    start=1
):

    print("=" * 60)
    print(f"Running CatBoost Balanced - Fold {fold}")


    # --------------------------------------------------------
    # 1. Split training fold dan validation fold
    # --------------------------------------------------------

    X_tr = X_train.iloc[train_idx]
    X_val = X_train.iloc[val_idx]

    y_tr = y_train.iloc[train_idx]
    y_val = y_train.iloc[val_idx]


    # --------------------------------------------------------
    # 2. Hitung class weight dari TRAINING FOLD saja
    # --------------------------------------------------------

    neg_fold = (y_tr == 0).sum()
    pos_fold = (y_tr == 1).sum()

    scale_ratio_fold = float(
        neg_fold / pos_fold
    )

    print(f"Negative class : {neg_fold}")
    print(f"Positive class : {pos_fold}")
    print(f"Class ratio    : {scale_ratio_fold:.4f}")


    # --------------------------------------------------------
    # 3. Buat CatBoost Balanced
    # --------------------------------------------------------

    catboost_model = create_model_pipeline(

        CatBoostClassifier(
            iterations=300,
            learning_rate=0.05,
            depth=6,
            class_weights=[
                1.0,
                scale_ratio_fold
            ],
            verbose=0,
            random_state=42
        ),

        use_smote=False,
        use_scaler=False
    )


    # --------------------------------------------------------
    # 4. FIT
    # --------------------------------------------------------

    catboost_model.fit(
        X_tr,
        y_tr
    )


    # --------------------------------------------------------
    # 5. Prediction
    # --------------------------------------------------------

    y_pred = catboost_model.predict(
        X_val
    ).ravel()

    y_prob = catboost_model.predict_proba(
        X_val
    )[:, 1]


    # --------------------------------------------------------
    # 6. Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_val,
        y_pred
    )

    precision = precision_score(
        y_val,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        y_pred,
        zero_division=0
    )

    # F2-Score
    f2 = fbeta_score(
        y_val,
        y_pred,
        beta=2,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_val,
        y_prob
    )

    pr_auc = average_precision_score(
        y_val,
        y_prob
    )


    # --------------------------------------------------------
    # 7. Simpan hasil fold
    # --------------------------------------------------------

    catboost_cv_results.append({

        "Fold": fold,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1": f1,

        "F2": f2,

        "ROC-AUC": roc_auc,

        "PR-AUC": pr_auc
    })


    # --------------------------------------------------------
    # 8. Print hasil fold
    # --------------------------------------------------------

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1       : {f1:.4f}")
    print(f"F2       : {f2:.4f}")
    print(f"ROC-AUC  : {roc_auc:.4f}")
    print(f"PR-AUC   : {pr_auc:.4f}")

    print(f"✓ Fold {fold} completed")


# ============================================================
# HASIL SETIAP FOLD
# ============================================================

catboost_cv_df = pd.DataFrame(
    catboost_cv_results
)

print("\n")
print("=" * 60)
print("CATBOOST BALANCED - 5 FOLD CV RESULTS")
print("=" * 60)

display(
    catboost_cv_df.round(4)
)
```

## Step 14 - Pemilihan Model dan Threshold Tuning

Berdasarkan hasil hyperparameter tuning menggunakan Cross-Validation dengan F2-Score sebagai objective, XGBoost - Balanced memiliki Best CV F2 sebesar 0.5536 dan dipilih sebagai model final.

```python
# ============================================================
# STEP 14 — MODEL SELECTION & THRESHOLD TUNING
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_predict
from sklearn.metrics import (
    precision_score,
    recall_score,
    fbeta_score
)

# ============================================================
# 1. SELECT FINAL MODEL
# ============================================================

selected_model = xgb_balanced_tuning.best_estimator_

print("=" * 60)
print("SELECTED MODEL")
print("=" * 60)
print("Model      : XGBoost - Balanced")
print("Best CV F2 :", xgb_balanced_tuning.best_score_)


# ============================================================
# 2. GENERATE OUT-OF-FOLD PREDICTIONS
# ============================================================

y_prob_oof = cross_val_predict(
    selected_model,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]


# ============================================================
# 3. THRESHOLD SEARCH
# ============================================================

thresholds = np.arange(0.10, 0.71, 0.01)

threshold_results = []

for threshold in thresholds:

    y_pred_oof = (
        y_prob_oof >= threshold
    ).astype(int)

    precision = precision_score(
        y_train,
        y_pred_oof,
        zero_division=0
    )

    recall = recall_score(
        y_train,
        y_pred_oof,
        zero_division=0
    )

    f2 = fbeta_score(
        y_train,
        y_pred_oof,
        beta=2,
        zero_division=0
    )

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F2": f2
    })


# ============================================================
# 4. CREATE THRESHOLD RESULT TABLE
# ============================================================

threshold_results_df = pd.DataFrame(threshold_results)

display(
    threshold_results_df.round(4)
)


# ============================================================
# 5. SELECT OPTIMAL THRESHOLD BASED ON F2
# ============================================================

best_threshold_row = threshold_results_df.loc[
    threshold_results_df["F2"].idxmax()
]

best_threshold = float(
    best_threshold_row["Threshold"]
)

best_f2 = float(
    best_threshold_row["F2"]
)

best_precision = float(
    best_threshold_row["Precision"]
)

best_recall = float(
    best_threshold_row["Recall"]
)


# ============================================================
# 6. DISPLAY OPTIMAL THRESHOLD
# ============================================================

print("=" * 60)
print("OPTIMAL THRESHOLD")
print("=" * 60)

print(f"Threshold : {best_threshold:.2f}")
print(f"Precision : {best_precision:.4f}")
print(f"Recall    : {best_recall:.4f}")
print(f"F2-Score  : {best_f2:.4f}")
```

## Step 15 - OOF Evaluation with Optimal Threshold

```python
# ============================================================
# STEP 15 — OOF EVALUATION WITH OPTIMAL THRESHOLD
#
# Model          : XGBoost - Balanced
# Data           : Out-of-Fold Prediction (X_train)
# Primary Metric : F2-Score
# Secondary      : Recall, Precision
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# ============================================================
# 1. APPLY OPTIMAL THRESHOLD
# ============================================================

y_pred_oof = (
    y_prob_oof >= best_threshold
).astype(int)


# ============================================================
# 2. CALCULATE OOF METRICS
# ============================================================

oof_accuracy = accuracy_score(
    y_train,
    y_pred_oof
)

oof_precision = precision_score(
    y_train,
    y_pred_oof,
    zero_division=0
)

oof_recall = recall_score(
    y_train,
    y_pred_oof,
    zero_division=0
)

oof_f1 = f1_score(
    y_train,
    y_pred_oof,
    zero_division=0
)

oof_f2 = fbeta_score(
    y_train,
    y_pred_oof,
    beta=2,
    zero_division=0
)

oof_roc_auc = roc_auc_score(
    y_train,
    y_prob_oof
)

oof_pr_auc = average_precision_score(
    y_train,
    y_prob_oof
)


# ============================================================
# 3. DISPLAY OOF RESULTS
# ============================================================

print("=" * 60)
print("STEP 15 — OOF EVALUATION")
print("=" * 60)

print(f"Model     : XGBoost - Balanced")
print(f"Threshold : {best_threshold:.2f}")
print("-" * 60)

print(f"Accuracy  : {oof_accuracy:.4f}")
print(f"Precision : {oof_precision:.4f}")
print(f"Recall    : {oof_recall:.4f}")
print(f"F1-Score  : {oof_f1:.4f}")
print(f"F2-Score  : {oof_f2:.4f}")
print(f"ROC-AUC   : {oof_roc_auc:.4f}")
print(f"PR-AUC    : {oof_pr_auc:.4f}")


# ============================================================
# 4. CONFUSION MATRIX
# ============================================================

cm_oof = confusion_matrix(
    y_train,
    y_pred_oof
)

print("\nConfusion Matrix:")
print(cm_oof)


# ============================================================
# 5. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_train,
        y_pred_oof,
        target_names=[
            "No Subscription",
            "Subscription"
        ],
        zero_division=0
    )
)
```

## Step 18 - Identify Contributing Factors Using SHAP

Tujuan: mengidentifikasi variabel yang paling berkontribusi terhadap prediksi subscription berdasarkan model final XGBoost - Balanced

```python
# ============================================================
# STEP 18 — IDENTIFY CONTRIBUTING FACTORS USING SHAP
#
# Final Model : XGBoost - Balanced
# Purpose     : Identify features contributing to prediction
# ============================================================

import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# 1. GET PREPROCESSOR AND XGBOOST MODEL
# ============================================================

preprocessor = final_model.named_steps["preprocessor"]
xgb_model = final_model.named_steps["model"]


# ============================================================
# 2. TRANSFORM TEST DATA
# ============================================================

X_test_transformed = preprocessor.transform(X_test)


# ============================================================
# 3. GET FEATURE NAMES
# ============================================================

feature_names = preprocessor.get_feature_names_out()

print("Number of transformed features:")
print(len(feature_names))


# ============================================================
# 4. CONVERT SPARSE MATRIX TO DENSE IF NECESSARY
# ============================================================

if hasattr(X_test_transformed, "toarray"):
    X_test_shap = X_test_transformed.toarray()
else:
    X_test_shap = X_test_transformed


# ============================================================
# 5. CREATE SHAP EXPLAINER
# ============================================================

explainer = shap.TreeExplainer(
    xgb_model
)

shap_values = explainer.shap_values(
    X_test_shap
)


# ============================================================
# 6. SHAP SUMMARY PLOT
# ============================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_test_shap,
    feature_names=feature_names,
    max_display=15
)
```
